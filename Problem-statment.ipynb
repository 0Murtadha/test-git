```python
import pandas as pd
import plotly.express as px
```

```python
df = pd.read_csv(r"D:\GA\DATA FRAMES\listings.csv")
```

```python
df.columns
```

```python
df['price'] = (df['price'].str.replace('$','',regex=False).str.replace(',', '', regex=True)).astype(float)
df['first_review']=pd.to_datetime(df['first_review'])
df['last_review']=pd.to_datetime(df['last_review'])
df['price_quote_checkin_date']=pd.to_datetime(df['price_quote_checkin_date'])
df['price_quote_checkout_date'] = pd.to_datetime(df['price_quote_checkout_date'])
```

```python
df.drop(columns= list(df.columns[df.isnull().all()]), inplace= True)
#useless data
df.drop(columns=['id','name', 'first_review', 'last_review','description','host_listings_count','number_of_reviews_l30d', 'availability_eoy', 'number_of_reviews_ly','estimated_occupancy_l365d', 'estimated_revenue_l365d','review_scores_accuracy','review_scores_cleanliness', 'review_scores_checkin','review_scores_communication', 'review_scores_location','review_scores_value', 'calculated_host_listings_count','calculated_host_listings_count_entire_homes','calculated_host_listings_count_private_rooms','calculated_host_listings_count_shared_rooms','maximum_nights_avg_ntm','minimum_nights_avg_ntm','maximum_maximum_nights','minimum_maximum_nights','maximum_minimum_nights','minimum_minimum_nights','price_quote_raw','price_quote_price_per_night','price_quote_total_price' ,'hosts_time_as_host_months','hosts_time_as_user_months','hosts_time_as_user_years','region_parent_id','region_parent_name','region_id','host_id','host_profile_id','host_location','host_name','listing_url','scrape_id','last_searched', 'last_scraped','source','picture_url','host_url','host_profile_url','host_about','host_picture_url','latitude','longitude','calendar_last_scraped','reviews_per_month'], inplace=True)
df.fillna({'has_availability':'f'}, inplace=True)
```

```python
df.reset_index(drop=True,inplace=True)
```

```python
dfhome = df[df['room_type'] == 'Entire home/apt']
```

```python
avdfhome = dfhome[dfhome['availability_365'] > 250]
```

```python
(avdfhome['availability_365'].count()/dfhome['availability_365'].count()).round(2)*100
```

```python
(avdfhome[avdfhome['price'] > 315.5]['price'].count()/avdfhome['price'].count()).round(3)*100
```

```python
avdfhome.groupby('region_name')['availability_365'].count()/dfhome.groupby('region_name')['availability_365'].count()*100
```

```python
fig = px.bar(avdfhome.groupby('region_name')['availability_365'].count()/dfhome.groupby('region_name')['availability_365'].count()*100)
fig.show()
fig.write_image("my_graph.png")
```

```python
avdfhome['price median'] = avdfhome['price'].map(lambda x: 'above median' if x  > 315.5 else 'below median')
pricev = avdfhome.groupby('price median')['price'].count()
fig = px.pie(pricev, values= pricev.values, names= pricev.index)
fig.show()
fig.write_image("my_1.png")
```

```python
dfhome['above 250'] = dfhome['availability_365'].map(lambda x: 'high availability' if x  > 250 else 'normal availability')
reviewv = dfhome.groupby('above 250')['review_score'].count()
fig = px.bar(reviewv, values= reviewv.values, names= reviewv.index)
fig.show()
fig.write_image("my_2.png")
```
